## 1. What is Encapsulation?

**Encapsulation** means:

> Keeping data and the methods that work on that data together inside a class, while controlling how the data is accessed or modified.

A class combines:

- **Data / attributes**
- **Methods / behavior**

Example:

```text
BankAccount
├── balance        ← data
├── deposit()      ← behavior
└── withdraw()     ← behavior
```

````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

account = BankAccount(5000)
account.deposit(1000)

print(account.balance)
````

### Why is this encapsulation?

The `balance` data and the `deposit()` behavior are kept together inside the `BankAccount` class.

However, `balance` is still public, so anyone can change it directly.

````python
account.balance = -100000

print(account.balance)
````

This is one reason we introduce **data hiding / controlled access**.

# 2. What is Data Hiding?

**Data hiding** means restricting or discouraging direct access to internal data.

The main goal is:

> Protect the internal state of an object and control how that state changes.

Python commonly uses three naming styles:

| Type | Syntax | Meaning |
|---|---|---|
| Public | `name` | Normal access from anywhere |
| Protected convention | `_name` | Internal use; subclasses may use it |
| Private / name-mangled | `__name` | Mainly class-internal |

# 3. Public Members

A **public member** can be accessed normally from:

- Inside the class
- A subclass
- Outside the class

Public attributes do not begin with an underscore.

````python
class Student:
    def __init__(self):
        self.name = "Ram"

student = Student()

print(student.name)

student.name = "Hari"
print(student.name)
````

### Public member idea

```text
self.name
   │
   ├── Same class    ✓
   ├── Subclass      ✓
   └── Outside code  ✓
```

Use public members when direct access is completely acceptable.

# 4. Protected Members in Python

A protected-style member begins with **one underscore**:

```python
_name
```

Example:

```python
self._salary
```

The underscore tells other developers:

> This is an internal implementation detail. You can technically access it, but you should normally not use it directly from outside.

Important:

**Python does not enforce protected access.**

The single underscore is mainly a **naming convention**.

````python
class Employee:
    def __init__(self):
        self._salary = 50000

employee = Employee()

# This still works in Python.
print(employee._salary)
````

### So why use `_salary`?

The `_` gives a hint to developers and tools such as IDEs or linters.

```text
_salary
   │
   ├── Same class    ✓ Intended
   ├── Subclass      ✓ Intended
   └── Outside code  ✓ Works, but discouraged
```

Protected-style members are especially useful with **inheritance**.

````python
class Employee:
    def __init__(self, salary):
        self._salary = salary


class Manager(Employee):
    def calculate_bonus(self):
        return self._salary * 0.10


manager = Manager(50000)

print(manager.calculate_bonus())
````

Here, `Manager` is a subclass of `Employee`.

The child class uses `_salary`, which is a common intended use of protected-style members.

# 5. Private Members

A private-style member begins with **two underscores**:

```python
__name
```

Example:

```python
self.__balance
```

Double underscore activates a Python feature called **name mangling**.

This makes accidental direct access from outside harder.

````python
class Account:
    def __init__(self):
        self.__balance = 5000

    def show_balance(self):
        print(self.__balance)


account = Account()

account.show_balance()
````

Trying to access `__balance` directly from outside normally fails:

```python
account.__balance
```

Python will usually raise:

```text
AttributeError
```

````python
try:
    print(account.__balance)
except AttributeError as e:
    print("AttributeError:", e)
````

# 6. Name Mangling

Python does not create a truly inaccessible private variable.

For:

```python
self.__balance
```

Python internally changes the name approximately to:

```python
_Account__balance
```

This is called **name mangling**.

````python
print(account.__dict__)
````

You may see something like:

```text
{'_Account__balance': 5000}
```

Technically, we can still access it using the mangled name:

````python
print(account._Account__balance)
````

But this should normally **not** be done.

### Important teaching point

`__name` in Python does **not** mean absolute security.

It mainly:

- avoids accidental access,
- helps prevent name conflicts in subclasses,
- signals that the member is class-internal.

# 7. Public vs Protected vs Private

Let's place all three in one example.

````python
class Employee:
    def __init__(self):
        self.name = "Ram"          # Public
        self._department = "IT"    # Protected convention
        self.__salary = 50000      # Private / name-mangled

    def show_salary(self):
        return self.__salary


employee = Employee()

print("Public:", employee.name)
print("Protected:", employee._department)
print("Private through method:", employee.show_salary())
````

### Summary

| Member | Example | Outside Access |
|---|---|---|
| Public | `name` | Normal |
| Protected | `_name` | Works, but discouraged |
| Private | `__name` | Direct access normally fails |

# 8. Data Hiding with Getter Methods

A **getter** is a method used to read an internal value.

Instead of:

```python
student.__age
```

we provide:

```python
student.get_age()
```

````python
class Student:
    def __init__(self, age):
        self.__age = age

    def get_age(self):
        return self.__age


student = Student(20)

print(student.get_age())
````

The object keeps the actual data internal, while the method provides controlled access.

# 9. Data Hiding with Setter Methods

A **setter** is a method used to change internal data.

The important advantage is that we can add **validation**.

````python
class Student:
    def __init__(self, age):
        self.__age = age

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age must be greater than 0")


student = Student(20)

student.set_age(25)
print(student.get_age())
````

Now try an invalid value:

````python
student.set_age(-100)

print(student.get_age())
````

The setter protects the object from an invalid change.

```text
Outside code
    │
    │ set_age(-100)
    ▼
Validation
    │
    ├── Valid   → update
    └── Invalid → reject
```

This is a central purpose of encapsulation:

> **Control how internal state changes.**

# 10. Data Hiding with `@property`

Python provides a more convenient way to create controlled attribute access using `@property`.

It allows us to call a method as if it were a normal attribute.

````python
class Student:
    def __init__(self, age):
        self.__age = age

    @property
    def age(self):
        return self.__age


student = Student(20)

print(student.age)
````

Notice that we write:

```python
student.age
```

instead of:

```python
student.get_age()
```

But internally, the `age()` method is executed.

# 11. `@property.setter`

We can also control assignment using a setter.

````python
class Student:
    def __init__(self, age):
        self.__age = age

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        if value > 0:
            self.__age = value
        else:
            print("Invalid age")


student = Student(20)

print(student.age)

student.age = 25
print(student.age)
````

Now try an invalid value:

````python
student.age = -10

print(student.age)
````

This looks like normal attribute access:

```python
student.age
student.age = 25
```

but Python is actually running getter and setter methods.

This is often more Pythonic than traditional `get_age()` / `set_age()` methods.

# 12. Read-Only Property

Sometimes we want external code to **read** a value but not modify it directly.

We can create a property without a setter.

````python
class Employee:
    def __init__(self, employee_id):
        self.__employee_id = employee_id

    @property
    def employee_id(self):
        return self.__employee_id


employee = Employee(101)

print(employee.employee_id)
````

Because there is no setter, assignment such as:

```python
employee.employee_id = 500
```

will fail.

````python
try:
    employee.employee_id = 500
except AttributeError as e:
    print("AttributeError:", e)
````

# 13. Public, Protected and Private Methods

The same naming conventions also apply to methods.

````python
class Demo:
    def public_method(self):
        print("Public method")

    def _protected_method(self):
        print("Protected-style method")

    def __private_method(self):
        print("Private/name-mangled method")

    def call_private_method(self):
        self.__private_method()


obj = Demo()

obj.public_method()
obj._protected_method()
obj.call_private_method()
````

### Method naming

| Method | Type |
|---|---|
| `show()` | Public |
| `_show()` | Protected convention |
| `__show()` | Private / name-mangled |

# 14. Complete Real-World Example: Bank Account

This example combines:

- Private data
- Getter-like property
- Controlled deposit
- Controlled withdrawal
- Validation

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print(f"Withdrawn: {amount}")


account = BankAccount("Ram", 5000)

print("Initial balance:", account.balance)

account.deposit(1000)
print("After deposit:", account.balance)

account.withdraw(2000)
print("After withdrawal:", account.balance)

account.withdraw(10000)
````

### Why is this better?

Instead of letting outside code directly change the balance:

```python
account.balance = -100000
```

we force changes to go through:

```python
account.deposit(...)
account.withdraw(...)
```

Those methods can validate the operation.

So the flow becomes:

```text
User
 │
 ├── deposit(amount)
 │       │
 │       └── validation
 │              │
 │              ▼
 │         __balance
 │
 └── withdraw(amount)
         │
         └── validation
                │
                ▼
           __balance
```

That is the real purpose of encapsulation and data hiding.

# 15. Encapsulation vs Data Hiding

These two terms are related but not exactly the same.

### Encapsulation

**Encapsulation is the broader OOP concept.**

It means:

> Bundle data and related methods inside a class and provide a controlled interface.

### Data Hiding

**Data hiding is one technique used inside encapsulation.**

It means:

> Restrict or discourage direct access to internal implementation details.

A useful relationship is:

```text
ENCAPSULATION
     │
     ├── Data + methods inside a class
     │
     ├── Controlled access
     │
     └── Data hiding
            │
            ├── _protected
            ├── __private
            ├── getter/setter
            └── @property
```

# 16. Important Python-Specific Notes

Remember these points:

1. Python does not have Java/C++ style enforced `public`, `protected`, and `private` keywords.
2. `name` is a normal public member.
3. `_name` is mainly a **developer convention** for internal/protected use.
4. Python usually still allows outside access to `_name`.
5. `__name` triggers **name mangling**.
6. Name mangling is not the same as absolute security.
7. Getter/setter methods can provide validation.
8. `@property` gives controlled access with clean Python syntax.

# 19. Practice Task

Create a `Product` class with:

- Public `name`
- Protected `_category`
- Private `__price`
- A property to read `price`
- A property setter that prevents negative prices

Example:

```python
product = Product("Laptop", "Electronics", 80000)

print(product.name)
print(product.price)

product.price = 90000
print(product.price)

product.price = -500
```

Try implementing it yourself before checking a solution.

````python
# Practice solution

class Product:
    def __init__(self, name, category, price):
        self.name = name
        self._category = category
        self.__price = price

    @property
    def price(self):
        return self.__price

    @price.setter
    def price(self, value):
        if value >= 0:
            self.__price = value
        else:
            print("Price cannot be negative")


product = Product("Laptop", "Electronics", 80000)

print(product.name)
print(product.price)

product.price = 90000
print(product.price)

product.price = -500
print(product.price)
````
