## Introduction

A **function** is a named block of code that performs a specific task. Instead of writing the same
instructions again and again, we write them once inside a function, and then use ("call") that function
whenever we need that task done.

A **main function** is simply a function, usually named `main()`, that a programmer uses to mark where
the program's actual work begins.

This topic matters because almost every real Python program — small or large — uses this pattern to stay
organized. As a beginner, you will use this pattern in almost every script you write from now on, so it
is worth understanding well from the start.


## Detailed Explanation

### 1. What a function is and why functions are used
A function is a block of code with a name. We use functions so we don't have to repeat the same steps
over and over — we write the steps once, and reuse them by simply calling the function's name.


````python
def greet(person_name):
    print("Hello,", person_name)

greet("Ram")
greet("Hari")
````

**Observation:** We wrote the greeting instructions only once, inside `greet()`, but used them twice —
once for Ram and once for Hari.


### 2. The purpose of a main function structure
Instead of writing all our program's logic loose at the top of the file, we place the main flow of the
program inside a function called `main()`. This clearly marks "this is where the program's work starts."


````python
def main():
    name = "Ram"
    print("Welcome,", name)

main()
````

**Observation:** All the program's activity happens inside `main()`. Anyone reading the file can look
for `main()` first to understand what the program actually does.


### 3. The role of function definition and function execution
There are two separate steps:
- **Defining** a function (writing `def function_name():` and its code) only teaches Python the steps.
  Nothing runs yet.
- **Executing** (calling) it — writing `function_name()` — is the moment the code inside actually runs.


````python
def show_message():
    print("This only prints when the function is CALLED.")

print("The function has been defined.")
show_message()   # this is the CALL - now it actually runs
````

**Observation:** Nothing printed from inside `show_message()` until we actually called it. Defining a
function is like writing a recipe; calling it is like actually cooking the recipe.


### 4. The meaning and purpose of the special main execution check
Every Python file has a hidden variable called `__name__`.
- If the file is run directly, Python sets `__name__` to the text `"__main__"`.
- If the file is instead imported by another file, `__name__` becomes something else (the file's name),
  NOT `"__main__"`.

So this check:
```
if __name__ == "__main__":
    main()
```
means: "Only run `main()` if this file was started directly — not if it was simply borrowed by another
file."


````python
def main():
    print("main() ran because this file was run directly.")

if __name__ == "__main__":
    main()

print("Current value of __name__:", __name__)
````

**Observation:** `__name__` shows `"__main__"` here because we ran this notebook directly, so the
`if` condition was `True` and `main()` ran.


### 5. Why organizing a program with a main function improves readability and structure
Using `main()` separates "defining tools" (functions) from "using the tools" (the actual program flow).
This makes a file easier to read, easier to test, and safer to reuse in other programs, since importing
it will not automatically run everything inside it.


````python
# Messy version - everything mixed together
print("--- Messy version ---")
name = "Hari"
age = 20
print(name, "is", age, "years old.")

# Organized version - using functions and main()
print("--- Organized version ---")

def describe_person(person_name, person_age):
    print(person_name, "is", person_age, "years old.")

def main():
    name = "Hari"
    age = 20
    describe_person(name, age)

main()
````

**Observation:** Both versions print the same result, but the organized version is easier to read and
easier to reuse, since `describe_person()` can now be called again for other people too.


## Multiple Experiments

### Experiment 1: Defining a Function Without Calling It

**What We Are Testing:** Whether writing a function alone is enough to make it run.


````python
def main():
    print("You will never see this message printed.")

print("Program finished without calling main().")
````

**Observation:** Because `main()` was never called, its code never ran, even though it was written
correctly. This shows that defining a function only teaches Python the steps — it does not run them.


### Experiment 2: Calling a Function Correctly

**What We Are Testing:** What changes when we actually call the function we forgot to call above.


````python
def main():
    print("Now you WILL see this message, because main() was called.")

main()
````

**Observation:** Adding the call `main()` at the bottom made the message appear. This is the fix for
Experiment 1's mistake.


### Experiment 3: Checking the Value of __name__

**What We Are Testing:** What `__name__` equals when a file (or notebook) is run directly.


````python
print(__name__)
````

**Observation:** It shows `"__main__"`, confirming this notebook is being run directly and not
imported by another file.


### Experiment 4: Simulating the "Imported vs Run Directly" Check

**What We Are Testing:** How the `if __name__ == "__main__":` check decides whether to run `main()`.


````python
def main():
    print("main() called!")

if __name__ == "__main__":
    main()
else:
    print("This file was imported, so main() was NOT called automatically.")
````

**Observation:** Because `__name__` is `"__main__"` here, the `if` branch ran. If this exact code
lived in a file called `helper.py` and another program did `import helper`, the `else` branch would run
instead, and `main()` would stay silent.


### Experiment 5: Calling main() More Than Once

**What We Are Testing:** What happens if `main()` is called two times instead of one.


````python
def main():
    print("main() is running.")

main()
main()
````

**Observation:** The message printed twice, once for each call. This shows a function runs fresh, in
full, every single time it is called.


### Experiment 6: Using main() to Call Multiple Helper Functions

**What We Are Testing:** How `main()` can organize several smaller functions into one program flow.


````python
def show_greeting():
    print("Hello, Sita!")

def show_goodbye():
    print("Goodbye, Sita!")

def main():
    show_greeting()
    show_goodbye()

main()
````

**Observation:** `main()` decided the ORDER in which the two helper functions ran. This is exactly why
`main()` is useful — it organizes the overall flow of the program.


## Important Rules
- Writing a function does not run it — you must CALL it.
- `__name__` equals `"__main__"` only when a file is run directly, not when it is imported.
- `main()` is just a name we choose by convention — Python does not require this exact name.
- Code placed inside `main()` only runs when `main()` is called.


## Common Mistakes

**Mistake 1: Forgetting to call main()**
```
def main():
    print("Hello!")
# main() is never called here
```
This is wrong because the function is defined but never executed, so nothing happens.
**How to avoid it:** Always add `main()` (or `if __name__ == "__main__": main()`) at the bottom of the
file.

**Mistake 2: Misspelling __main__**
```
if __name__ == "main":   # missing double underscores
```
This is wrong because `__name__` will never equal the plain text `"main"` — it will be `"__main__"`. The
condition will always be `False`, silently skipping the code inside.
**How to avoid it:** Always type `__main__` with two underscores on each side.

**Mistake 3: Writing important logic outside any function**
```
name = "Ram"
print(name)
```
This is wrong (for larger programs) because it runs immediately for anyone who imports the file, even if
they only wanted one small function from it.
**How to avoid it:** Place program logic inside `main()` and only run it using the `if __name__ ==
"__main__":` check.
