## Introduction

A **string** is Python's way of representing text — anything from a single character to a whole
paragraph. Strings are one of the most frequently used data types in programming, since almost every
program eventually needs to display messages, read text, or process words in some way.


## Detailed Explanation

### 1. What strings are and how text is represented
Any sequence of characters wrapped in quotes is treated as text (a string) by Python.


````python
greeting = "Namaste"
print(greeting)
print(type(greeting))
````

**Observation:** `greeting` holds text, and `type()` confirms it is a `str`.


### 2. Different ways of creating text values
Single quotes, double quotes, and triple quotes all create strings; triple quotes additionally allow text
to span multiple lines.


````python
single = 'Ram'
double = "Hari"
multiline = """This is
a two line message."""

print(single)
print(double)
print(multiline)
````

**Observation:** All three quoting styles created valid strings; only triple quotes let the text span
more than one line.


### 3. String length
`len(text)` counts every character in a string, including spaces.


````python
name = "Ram Kumar"
print(len(name))
````

**Observation:** The space between "Ram" and "Kumar" is ALSO counted as a character.


### 4. Character access and indexing
Each character has a position, called an index, starting from `0` for the first character.


````python
word = "Python"
print(word[0])   # first letter
print(word[5])   # last letter
````

**Observation:** `"Python"` has 6 letters, at positions 0 through 5, so position 5 gives the last
letter.


### 5. Positive and negative indexing
Negative indexes count backward from the end, with `-1` as the last character.


````python
word = "Python"
print(word[-1])    # last letter
print(word[-6])    # first letter
````

**Observation:** `word[-1]` and `word[5]` both give the same letter, just counted from opposite
directions.


### 6. String slicing
Slicing extracts a portion of a string using a start position and an end position. The end position is
NOT included in the result.


````python
sentence = "Learning Python is fun"
print(sentence[0:8])
print(sentence[9:])
print(sentence[::-1])   # reversed
````

**Observation:** `sentence[0:8]` stops right BEFORE index 8, and `sentence[::-1]` reverses the whole
string using a step of `-1`.


### 7. Important string operations and methods
Strings come with many built-in methods, like changing case, removing extra spaces, replacing text, and
splitting text into pieces.


````python
text = "  Hello Python World  "

print(text.upper())
print(text.lower())
print(text.strip())
print(text.strip().replace("Python", "Coding"))
print(text.strip().split(" "))
````

**Observation:** Each method returns a NEW string with a specific change applied, without changing
the original `text` variable itself.


### 8. The immutable nature of strings
Once made, a string cannot be changed directly. Any "change" actually creates a brand-new string.


````python
word = "Python"
try:
    word[0] = "J"
except TypeError as error:
    print("Error:", error)

# The only way to "change" it is to make a new string:
word = "J" + word[1:]
print(word)
````

**Observation:** Directly assigning to a character position always raises a `TypeError`. The only way
to get a changed result is to build a brand-new string.


## Multiple Experiments

### Experiment 1: Accessing an Index That Doesn't Exist

**What We Are Testing:** What error occurs when asking for a position beyond the string's length.


````python
word = "Python"
try:
    print(word[10])
except IndexError as error:
    print("Error:", error)
````

**Observation:** `"Python"` only has 6 letters (positions 0 to 5), so asking for position 10 raises an
`IndexError`.


### Experiment 2: Slicing With a Step Value

**What We Are Testing:** How a step value changes which characters are included in a slice.


````python
sentence = "abcdefghij"
print(sentence[::2])    # every second character
print(sentence[1::2])    # every second character, starting from index 1
````

**Observation:** Changing the step value (and the starting index) changes exactly which characters
are picked from the string.


### Experiment 3: Checking if a Word Exists Inside a Sentence

**What We Are Testing:** Whether the `in` keyword can search inside a string.


````python
sentence = "Python is a powerful language"
print("powerful" in sentence)
print("Java" in sentence)
````

**Observation:** `in` returns `True` when the given text is found somewhere inside the string, and
`False` otherwise.


### Experiment 4: Comparing .upper() and .lower() on the Same Text

**What We Are Testing:** Whether changing case affects the original variable.


````python
name = "Sita"
print(name.upper())
print(name.lower())
print(name)
````

**Observation:** `name` itself never changed — `.upper()` and `.lower()` each returned a NEW string
without touching the original.


### Experiment 5: Splitting a Sentence Into Words

**What We Are Testing:** How `.split()` breaks a sentence apart.


````python
sentence = "Ram likes to play football"
words = sentence.split(" ")
print(words)
print(len(words))
````

**Observation:** `.split(" ")` broke the sentence into a list of separate words, using the space as
the dividing point.


### Experiment 6: Changing the Slice Range and Observing the Result

**What We Are Testing:** How the result of a slice changes as we adjust the start and end numbers.


````python
word = "Programming"
print(word[0:4])
print(word[0:7])
print(word[4:11])
````

**Observation:** Increasing the end number includes more characters in the result, and moving the
start number forward skips more characters from the beginning.


## Important Rules
- Counting starts at 0, not 1.
- Slicing's ending number is NOT included in the result.
- Strings CANNOT be changed directly — string methods always give you a new string.
- `len()` counts every character, including spaces.


## Common Mistakes

**Mistake 1: Thinking the first letter is at position 1**
```
word[1]   # this is actually the SECOND letter, not the first
```
This is wrong because Python indexing starts at 0.
**How to avoid it:** Always remember `word[0]` is the first character.

**Mistake 2: Forgetting the slice's end index is excluded**
```
word[0:3]   # gives 3 characters, NOT 4
```
This is wrong if the beginner expects the character AT index 3 to be included — it is not.
**How to avoid it:** Remember the ending number is always excluded from the slice.

**Mistake 3: Trying to change a character directly**
```
word[0] = "J"
```
This is wrong because strings are immutable and cannot be changed in place.
**How to avoid it:** Build a new string instead, e.g. `word = "J" + word[1:]`.
