# Exception Handling in Python

A complete, hands-on guide to handling errors in Python.

### 1. What is Exception Handling?

An **exception** is an error that occurs during program execution and disrupts the normal flow of instructions. Exception handling lets you *catch* these errors and respond to them instead of letting the program crash.

Python uses four keywords for this:
- **try** – code that might raise an error
- **except** – runs if an error occurs
- **else** – runs if no error occurs
- **finally** – always runs, error or not (cleanup code)

```python
# Without exception handling — this crashes the program
num = int("abc")  # ValueError
```

### 2. Basic try / except

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ValueError:
    print("That's not a valid number!")
except ZeroDivisionError:
    print("You can't divide by zero!")
else:
    print(f"Result: {result}")
finally:
    print("Execution finished.")
```

### 3. Catching Any Exception

`except Exception as e` catches almost any error. Useful for debugging, but less precise than catching specific error types — use it sparingly.

```python
try:
    1 / 0
except Exception as e:
    print(f"Something went wrong: {e}")
```

### 4. The `else` Clause

`else` runs only if the `try` block completes **without** raising an exception.

```python
try:
    x = 10 / 2
except ZeroDivisionError:
    print("Division failed")
else:
    print(f"Success! x = {x}")
```

### 5. The `finally` Clause

`finally` always executes — whether an exception occurred or not. It's commonly used for cleanup (closing files, releasing resources, etc.).

```python
try:
    f = open("sample.txt", "w")
    f.write("Hello, World!")
except IOError:
    print("File error occurred")
finally:
    f.close()
    print("File closed.")
```

### 6. Nested try / except

`try` blocks can be nested inside each other for finer-grained error handling.

```python
try:
    try:
        num = int("not_a_number")
    except ValueError:
        print("Inner block: invalid conversion")
        raise  # re-raise to outer block
except ValueError:
    print("Outer block: caught the re-raised error")
```

### 7. Common Built-in Exceptions

| Exception | Meaning |
|---|---|
| `ValueError` | Invalid value (e.g. `int('abc')`) |
| `TypeError` | Wrong data type used |
| `ZeroDivisionError` | Division by zero |
| `IndexError` | Invalid list/sequence index |
| `KeyError` | Invalid dictionary key |
| `FileNotFoundError` | File does not exist |
| `AttributeError` | Invalid attribute reference |
| `ImportError` | Module import failed |

An example for each exception below:

**ValueError** – Invalid value

```python
try:
    num = int("abc")
except ValueError as e:
    print(f"ValueError: {e}")
```

**TypeError** – Wrong data type used

```python
try:
    result = "5" + 5
except TypeError as e:
    print(f"TypeError: {e}")
```

**ZeroDivisionError** – Division by zero

```python
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print(f"ZeroDivisionError: {e}")
```

**IndexError** – Invalid list/sequence index

```python
try:
    my_list = [1, 2, 3]
    print(my_list[10])
except IndexError as e:
    print(f"IndexError: {e}")
```

**KeyError** – Invalid dictionary key

```python
try:
    my_dict = {"a": 1}
    print(my_dict["b"])
except KeyError as e:
    print(f"KeyError: {e}")
```

**FileNotFoundError** – File does not exist

```python
try:
    with open("no_such_file.txt") as f:
        content = f.read()
except FileNotFoundError as e:
    print(f"FileNotFoundError: {e}")
```

**AttributeError** – Invalid attribute reference

```python
try:
    num = 10
    num.append(5)
except AttributeError as e:
    print(f"AttributeError: {e}")
```

**ImportError** – Module import failed

```python
try:
    import non_existing_module
except ImportError as e:
    print(f"ImportError: {e}")
```

### 8. Using `with` for Automatic Cleanup

Context managers (`with` statement) handle cleanup automatically — often a cleaner alternative to `try/finally` for resources like files.

```python
with open("sample.txt", "w") as f:
    f.write("Using context manager, no need for manual finally!")

print("File written and closed automatically.")
```

### 9. Summary

- Use `try/except` to prevent crashes and handle errors gracefully.
- Use `else` for code that should run only when no error occurs.
- Use `finally` for guaranteed cleanup code.
- Catch **specific** exceptions when possible; avoid bare `except:`.
- Prefer `with` statements for resource management when available.
