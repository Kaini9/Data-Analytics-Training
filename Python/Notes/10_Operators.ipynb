## Introduction

**Operators** are special symbols that perform an operation on values, called operands. Python groups
operators into several families based on what they do: calculating numbers, updating variables, comparing
values, combining logical conditions, checking identity, and checking membership.


## Detailed Explanation

### 1. Arithmetic operators
These perform mathematical calculations: `+` add, `-` subtract, `*` multiply, `/` divide, `//` floor
divide (drops the remainder), `%` modulus (gives just the remainder), `**` exponent (raises to a power).


````python
a, b = 17, 5

print("Add:", a + b)
print("Subtract:", a - b)
print("Multiply:", a * b)
print("Divide:", a / b)
print("Floor divide:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
````

**Observation:** `/` always gives a decimal answer. `//` gives only the whole-number part. `%` gives
just the leftover remainder.


### 2. Assignment operators
The basic `=` stores a value. Compound assignment operators (`+=`, `-=`, `*=`, and so on) combine a
calculation with reassignment in one step.


````python
score = 10
print("Start:", score)

score += 5    # same as score = score + 5
print("After +=5:", score)

score -= 3     # same as score = score - 3
print("After -=3:", score)

score *= 2      # same as score = score * 2
print("After *=2:", score)
````

**Observation:** Each compound operator performed a calculation AND reassigned the result back to the
same variable, all in one shorter line.


### 3. Comparison operators
These compare two values and always produce a Boolean result: `==` equal, `!=` not equal, `>` greater
than, `<` less than, `>=`, `<=`.


````python
x, y = 8, 12

print(x == y)
print(x != y)
print(x > y)
print(x < y)
````

**Observation:** Every comparison produces exactly one of two possible results: `True` or `False`.


### 4. Logical operators
`and` — True only if BOTH conditions are True. `or` — True if AT LEAST ONE condition is True. `not` —
flips True to False, or False to True.


````python
age = 20
has_id = True

print(age >= 18 and has_id)
print(age >= 18 or has_id)
print(not has_id)
````

**Observation:** `and` needed both sides to be True, `or` only needed one side, and `not` simply
flipped the value.


### 5. Identity operators
`is` and `is not` check whether two variables refer to the exact SAME object in memory — this is
different from checking whether their VALUES are equal.


````python
list_a = [1, 2, 3]
list_b = [1, 2, 3]
list_c = list_a

print(list_a == list_b)     # same VALUES? Yes
print(list_a is list_b)      # same OBJECT? No - two separate lists
print(list_a is list_c)      # same OBJECT? Yes - list_c points to list_a
````

**Observation:** `list_a` and `list_b` LOOK the same (`==` is True), but they are two separate boxes
(`is` is False). `list_c` is literally pointing to the SAME box as `list_a` (`is` is True).


### 6. Membership operators
`in` and `not in` check whether a value exists (or does not exist) within a collection, like a list or a
string.


````python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)
print("mango" in fruits)
print("mango" not in fruits)
````

**Observation:** `in` looks through the whole list checking for a match, and `not in` gives the
opposite answer.


## Multiple Experiments

### Experiment 1: Checking If a Number is Even or Odd

**What We Are Testing:** Using `%` to determine whether a number divides evenly by 2.


````python
number = 14
print(number % 2)

number = 15
print(number % 2)
````

**Observation:** A remainder of `0` means the number is even; a remainder of `1` means it is odd.

**What We Are Testing:** Whether `+=` can be used repeatedly to accumulate a running total.


````python
total = 0
total += 10
total += 20
total += 5
print("Final total:", total)
````

**Observation:** Each `+=` added to the running total, showing how compound assignment is useful for
building up a value step by step.


### Experiment 3: Combining Comparisons With Logical Operators

**What We Are Testing:** Whether we can check a range using `and`.


````python
marks = 75
print(marks >= 60 and marks <= 100)
````

**Observation:** Both conditions had to be True for the overall result to be True, correctly

### Experiment 4: is vs == With Numbers

**What We Are Testing:** Whether `is` and `==` always agree, even for simple values like numbers.


````python
x = 5
y = 5
print(x == y)
print(x is y)
````

**Observation:** For small numbers like this, `is` and `==` happen to agree, but this is not

### Experiment 5: Checking Membership in a String

**What We Are Testing:** Whether `in` works on strings the same way it works on lists.


````python
sentence = "Python is fun to learn"
print("fun" in sentence)
print("boring" in sentence)
````

**Observation:** `in` searched inside the string's characters for the given text, just like it

### Experiment 6: A Discount Check Using and / or

**What We Are Testing:** Combining `and` and `or` to decide if someone qualifies for a discount.


````python
age = 65
has_membership = False

eligible = age >= 60 or has_membership
print("Eligible for discount:", eligible)
````

**Observation:** Because the age condition alone was True, `or` made the whole expression True, even
though `has_membership` was False.


## Important Rules
- `/` always gives a decimal answer, even for evenly divisible numbers.
- `=` assigns a value, `==` compares two values — never confuse them.
- `is` checks identity (same object); `==` checks value.
- `and`, `or`, and `not` only make sense when combining or reversing Boolean expressions.


## Common Mistakes

**Mistake 1: Using = when == was intended**
```
if age = 18:
```
This is wrong because `=` is for assignment, not comparison — this actually causes a `SyntaxError`.
**How to avoid it:** Use `==` for comparisons: `if age == 18:`.

**Mistake 2: Assuming is and == always agree**
```
[1, 2] is [1, 2]   # False, even though the values look equal
```
This is wrong because `is` checks if they are the SAME object, not just equal values.
**How to avoid it:** Use `==` when you only care about the values being equal.

**Mistake 3: Mixing up % and //**
```
17 // 5   # gives 3 (the whole number result)
17 % 5     # gives 2 (the remainder)
```
This is wrong if the beginner confuses which one gives the remainder and which gives the whole result.
**How to avoid it:** Remember `%` is for the remainder, `//` is for the whole-number quotient.
