## Introduction

When an expression contains more than one operator, Python must decide which operation to perform first.
**Operator precedence** is the set of rules that determines this order — just like in math class, where
multiplication happens before addition unless parentheses say otherwise.


## Detailed Explanation

### 1. What operator precedence means
Precedence is a ranking: operators with higher precedence are evaluated before operators with lower
precedence, no matter the order they appear left to right in the expression.


````python
result = 2 + 3 * 4
print(result)
````

**Observation:** Even though `2 + 3` appears first when reading left to right, Python evaluated
`3 * 4` first because multiplication has higher precedence, giving `2 + 12 = 14`.


### 2. Why the order of operations matters
The same numbers and operators can produce completely different results depending on which operation
happens first.


````python
print(2 + 3 * 4)
print((2 + 3) * 4)
````

**Observation:** The exact same numbers gave two very different answers, `14` and `20`, purely
because of the order operations were performed in.


### 3. The general priority of parentheses, arithmetic operations, comparisons, and logical
operations
1. **Parentheses `()`** — always evaluated first.
2. **Exponentiation `**`**
3. **Multiplication, division, floor division, modulus (`*`, `/`, `//`, `%`)**
4. **Addition and subtraction (`+`, `-`)**
5. **Comparison operators (`==`, `!=`, `>`, `<`, `>=`, `<=`)**
6. **Logical `not`, then `and`, then `or`**


````python
print(2 * 3 ** 2)      # exponent first: 3**2=9, then 2*9=18
print(3 + 2 > 4)        # arithmetic first: 3+2=5, then 5>4
print(10 >= 18 and True) # comparison first: 10>=18 is False, then False and True
````

**Observation:** In every line, Python followed the priority order shown above — arithmetic before
comparisons, and comparisons before logical operators.


### 4. How parentheses can change the result of an expression
Because parentheses have the highest precedence, wrapping part of an expression in them forces that part
to be evaluated first, overriding the default order.


````python
print(2 * 3 + 1)
print(2 * (3 + 1))
````

**Observation:** Adding parentheses around `3 + 1` completely changed the answer from `7` to `8`,
even though the exact same numbers and operators were used.


## Multiple Experiments

### Experiment 1: Comparing Expressions With and Without Parentheses

**What We Are Testing:** How much difference parentheses make to the same set of numbers.


````python
print(10 - 2 * 3)
print((10 - 2) * 3)
````

**Observation:** Without parentheses, multiplication happened first, giving `4`. With parentheses,
subtraction happened first, giving `24` — a very large difference from one small change.


### Experiment 2: Exponent vs Multiplication

**What We Are Testing:** Whether `**` is evaluated before or after `*`.


````python
print(2 * 3 ** 2)
print((2 * 3) ** 2)
````

**Observation:** `**` runs before `*` by default, so the first line calculates `3**2=9` before
multiplying. Adding parentheses forces the multiplication to happen first instead.


### Experiment 3: Arithmetic vs Comparison

**What We Are Testing:** Whether math is always finished before a comparison is made.


````python
print(3 + 2 > 4)
print(3 + (2 > 4))
````

**Observation:** In the first line, `3 + 2` finishes before comparing to `4`. In the second line, the
parentheses force the comparison `2 > 4` to happen first, producing `False`, which then gets added to `3`
as if it were `0`.


### Experiment 4: Logical Operators and Comparisons Together

**What We Are Testing:** Whether comparisons finish before `and` combines them.


````python
age = 20
has_ticket = False

print(age >= 18 and has_ticket)
````

**Observation:** Python first evaluated `age >= 18` (True), then combined it with `has_ticket` using
`and`, giving `False` overall since `has_ticket` was False.


### Experiment 5: Breaking Down a Longer Expression Step by Step

**What We Are Testing:** Whether following the precedence rules by hand matches Python's actual answer.


````python
# 10 - 2 * 3 ** 2 + 4 / 2
# Step 1: 3 ** 2 = 9
# Step 2: 2 * 9 = 18
# Step 3: 4 / 2 = 2.0
# Step 4: 10 - 18 + 2.0 = -6.0

result = 10 - 2 * 3 ** 2 + 4 / 2
print(result)
````

**Observation:** Working through the steps by hand, following the priority order, matched exactly
what Python calculated.


### Experiment 6: Using Parentheses to Make a Condition Clearer

**What We Are Testing:** Whether adding parentheses to a logical expression changes its meaning or just
its readability.


````python
age = 20
has_id = True
has_ticket = False

result1 = age >= 18 and has_id or has_ticket
result2 = age >= 18 and (has_id or has_ticket)

print(result1)
print(result2)
````

**Observation:** Both expressions gave `True` here, but adding parentheses made the SECOND version's
intention much clearer to a human reader, even when the final answer happens to match.


## Important Rules
- Parentheses always have the highest precedence and can be used to force any desired order.
- Exponentiation runs before multiplication/division, which run before addition/subtraction.
- All arithmetic finishes before any comparison is made.
- All comparisons finish before logical operators (`and`, `or`) combine them.


## Common Mistakes

**Mistake 1: Assuming Python always reads strictly left to right**
```
2 + 3 * 4   # is 14, NOT 20
```
This is wrong if the beginner expects left-to-right reading — multiplication runs first regardless of
position.
**How to avoid it:** Learn and apply the priority order, or use parentheses to be explicit.

**Mistake 2: Forgetting and binds tighter than or**
```
True or False and False   # this is True, not False
```
This is wrong if the beginner expects strict left-to-right combination — `and` is evaluated before `or`.
**How to avoid it:** Use parentheses to make the intended grouping completely clear.

**Mistake 3: Not using parentheses when the order is confusing**
```
result = a + b * c - d / e
```
This is wrong (or at least risky) because it can be hard for a human reader to quickly confirm the
intended order.
**How to avoid it:** Add parentheses even when not strictly required, purely to improve readability.
