## Introduction

Some characters cannot simply be typed into a string because they either have no visible symbol (like a
new line) or would be confused with the string's own quote marks. **Escape characters** solve this: a
backslash `\` followed by a specific character tells Python to treat the next character specially.


## Detailed Explanation

### 1. What escape characters are and why they are needed
An escape sequence begins with a backslash `\` followed by another character, and Python interprets this
combination as a single special instruction rather than literal text.


````python
print("Ram says: hi\nHari says: hello")
````

**Observation:** The `\n` inside the string was not printed as two characters — instead it created
an actual line break in the output.


### 2. Creating new lines and tabs
`\n` inserts a new line. `\t` inserts a horizontal tab, similar to pressing the Tab key.


````python
print("Line one\nLine two\nLine three")
print("Name:\tRam")
print("Age:\t20")
````

**Observation:** `\n` moved each part of the text to its own new line, and `\t` aligned the text
after it like a small column.


### 3. Displaying quotation marks inside text
A quotation mark matching the string's own quote type can be shown using `\"` or `\'`. Alternatively,
you can simply switch to the OTHER quote type to avoid escaping entirely.


````python
print("Hari said, \"Python is fun!\"")

# Easier way - just switch quote types:
print('Hari said, "Python is fun!"')
````

**Observation:** Both lines print the exact same message, but switching quote styles avoided the
need for any escaping at all.


### 4. Displaying backslashes
Since a backslash starts an escape sequence, displaying a literal backslash requires typing it TWICE:
`\\`.


````python
print("The file path is C:\\Users\\Ram")
````

**Observation:** Each `\\` in the code became just ONE backslash character in the actual printed
output.


### 5. Combining multiple escape characters
Escape sequences can be freely combined within a single string to build more complex, formatted text.


````python
report = "Report:\n\tName:\tSita\n\tScore:\t95"
print(report)
````

**Observation:** New lines and tabs were combined together to produce a small, neatly indented report
using just one string.


## Multiple Experiments

### Experiment 1: Using \n to Print a Numbered List

**What We Are Testing:** Whether `\n` can be used to build a multi-line list in a single print statement.


````python
print("1. Wake up\n2. Study Python\n3. Practice coding")
````

**Observation:** Each numbered item appeared on its own line, all from one single `print()` call.


### Experiment 2: Comparing a Single Backslash to a Double Backslash

**What We Are Testing:** What happens when we forget to double the backslash in a file path.


````python
single_backslash_text = "C:\new\test"
double_backslash_text = "C:\\new\\test"

print(single_backslash_text)
print(double_backslash_text)
````

**Observation:** With a single backslash, `\n` inside the path was read as a real escape
character (a new line), so the path printed incorrectly, split across two lines instead of staying as
one line of text. Doubling the backslash (`\\`) fixed this by making it a literal backslash instead.


### Experiment 3: Using Raw Strings to Avoid Escaping

**What We Are Testing:** Whether prefixing a string with `r` stops escape sequences from being processed.


````python
normal_string = "C:\new\test"
raw_string = r"C:\new\test"

print(normal_string)
print(raw_string)
````

**Observation:** The raw string treats every backslash literally, which is why it is often used for
file paths instead of manually doubling every backslash.


### Experiment 4: Quotes Inside Quotes, Two Ways

**What We Are Testing:** Comparing escaping a quote versus switching quote styles.


````python
print("She said, \'hello\' to me.")
print('She said, \'hello\' to me.')
````

**Observation:** Escaping works inside either quote style, but it is only NECESSARY when the escaped
quote matches the string's own surrounding quote type.


### Experiment 5: A Multi-Line Report Using Both \n and \t

**What We Are Testing:** Combining escape characters to format a small student report.


````python
report = "Student Report\n\tName:\tShyam\n\tGrade:\tA\n\tRank:\t1"
print(report)
````

**Observation:** Combining `\n` and `\t` created a neatly structured multi-line, indented report from
one single string.


### Experiment 6: What Happens Without Any Escape Characters

**What We Are Testing:** Comparing text with and without escape characters to see the difference clearly.


````python
without_escapes = "Name: Ram Age: 20 City: Kathmandu"
with_escapes = "Name: Ram\nAge: 20\nCity: Kathmandu"

print(without_escapes)
print("---")
print(with_escapes)
````

**Observation:** Without escape characters, everything ran together on one line. Adding `\n`
organized the same information onto separate lines, much easier to read.


## Important Rules
- `\n` = new line, `\t` = tab.
- To show a real backslash, you need TWO backslashes: `\\`.
- `r"..."` makes a raw string where backslashes are treated normally, not as escape characters.
- Switching quote styles can often avoid the need to escape a quote at all.


## Common Mistakes

**Mistake 1: Forgetting to escape a matching quote**
```
print("She said, "hello"")
```
This is wrong because the second `"` ends the string early, causing a `SyntaxError`.
**How to avoid it:** Escape it (`\"`) or switch to single quotes for the outer string.

**Mistake 2: Using a single backslash in a file path**
```
"C:\Users\test"
```
This is wrong because some combinations (like `\U` or `\t`) may be interpreted as escape sequences
instead of literal characters.
**How to avoid it:** Double the backslashes (`\\`) or use a raw string (`r"..."`).

**Mistake 3: Confusing a raw string with a normal string**
```
print(r"Line1\nLine2")   # prints literally, no new line created
```
This is wrong if the beginner expected `\n` to create a new line — inside a raw string, it does not.
**How to avoid it:** Only use raw strings when you specifically want backslashes treated literally.
