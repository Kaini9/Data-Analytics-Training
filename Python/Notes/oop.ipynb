## 1. What is OOP?

**OOP** stands for **Object-Oriented Programming**.

Python is an object-oriented language. OOP organizes a program around **classes** and **objects** so that related data and behavior can be kept together.

### Advantages of OOP

According to the PDF, OOP:

- provides a clear structure to programs,
- makes code easier to maintain, reuse, and debug,
- helps keep code **DRY — Don't Repeat Yourself**,
- allows reusable applications to be built with less repeated code.

### DRY principle

The **DRY** idea means we should avoid writing the same logic again and again. Repeated behavior can be placed inside functions or classes and then reused.

## 3. The `__init__()` Method

Every class can define an `__init__()` method.

It is executed automatically when a new object is created.

Its main purpose is to give an object its **initial property values** and perform setup needed at object creation time.

### Basic pattern

```python
class ClassName:
    def __init__(self, value):
        self.value = value
```

When we call `ClassName(...)`, Python runs `__init__()` for the new object.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Emil", 36)

print(p1.name)
print(p1.age)
````

### Why use `__init__()`?

Without `__init__()`, properties often have to be assigned manually after each object is created.

Using `__init__()` makes object creation cleaner because the values are supplied at the moment the object is created.

````python
# Without __init__(): values are assigned manually
class PersonWithoutInit:
    pass

p1 = PersonWithoutInit()
p1.name = "Tobias"
p1.age = 25

print(p1.name)
print(p1.age)
````

````python
# With __init__(): values are supplied immediately
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Linus", 20)

print(p1.name)
print(p1.age)
````

### Default values in `__init__()`

A parameter can have a default value.

If the caller does not supply that argument, the default is used.

````python
class Person:
    def __init__(self, name, age=18):
        self.name = name
        self.age = age

p1 = Person("Emil")
p2 = Person("Tobias", 25)

print(p1.name, p1.age)
print(p2.name, p2.age)
````

### Multiple parameters

## 4. The `self` Parameter

`self` refers to the **current object**.

It is used inside instance methods to access properties and other methods that belong to that particular object.

For example:

```python
self.name
```

means:

> the `name` property of the object currently using this method.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print("Hello, my name is " + self.name)

p1 = Person("Emil", 25)
p1.greet()
````

### Why is `self` important?

Suppose two objects come from the same class:

- one object has the name `"Tobias"`,
- another has the name `"Linus"`.

The same method can be called by both objects. `self` tells Python **which object's data** the method should work with.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def printname(self):
        print(self.name)

p1 = Person("Tobias")
p2 = Person("Linus")

p1.printname()
p2.printname()
````

object.property

## 6. Methods in a Class

Methods are functions defined inside a class. They describe what objects created from that class can do.

> **Terminology note:** the PDF labels this section **"Class Methods"**, but its example is an ordinary method called through an object. This notebook keeps the PDF's topic while describing the behavior shown in the example.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello, my name is " + self.name)

p1 = Person("Emil")
p1.greet()
````

### Methods with parameters

Methods can receive extra arguments just like regular functions.

`self` identifies the current object; other parameters provide values needed for the operation.

````python
class Calculator:
    def add(self, a, b):
        return a + b

    def multiply(self, a, b):
        return a * b

calc = Calculator()

print(calc.add(5, 3))
print(calc.multiply(4, 7))
````

### Deleting a method from a class

The PDF also demonstrates removing a method from a class with `del`.

After the method is deleted from the class, objects of that class can no longer call it.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello!")

p1 = Person("Emil")
p1.greet()

del Person.greet

try:
    p1.greet()
except AttributeError as error:
    print("After deletion:", error)
````

# 7. Inheritance

**Inheritance** lets one class reuse behavior from another class.

The existing class is usually called the **parent/base class**, while the new class is called the **child/derived class**.

The inheritance diagram in the PDF lists:

- Single inheritance
- Multilevel inheritance
- Multiple inheritance
- Hierarchical inheritance
- Hybrid inheritance

## 7.1 Single Inheritance

**One parent → one child**

The child class inherits behavior from one parent class.

````python
class Parent:
    def show(self):
        print("Parent class")

class Child(Parent):
    def display(self):
        print("Child class")

c = Child()
c.show()
c.display()
````

## 7.2 Multilevel Inheritance

**Parent → child → grandchild**

A class inherits from a class that already inherited from another class.

The last class in the chain can access inherited behavior from the earlier levels.

````python
class Grandparent:
    def house(self):
        print("Grandparent house")

class Parent(Grandparent):
    def car(self):
        print("Parent car")

class Child(Parent):
    def bike(self):
        print("Child bike")

c = Child()
c.house()
c.car()
c.bike()
````

## 7.3 Multiple Inheritance

**Multiple parents → one child**

A child class inherits from more than one parent class.

````python
class Father:
    def skill(self):
        print("Father skill")

class Mother:
    def skill(self):
        print("Mother skill")

class Child(Father, Mother):
    pass

c = Child()
c.skill()
````

Main Problem: Same Method in Both Parents

````python
class Father:
    def show(self):
        print("Father show")

class Mother:
    def show(self):
        print("Mother show")

class Child(Father, Mother):
    pass
````

Python has to decide.

It uses MRO — Method Resolution Order.

# . Method Resolution Order (MRO)

The PDF introduces **MRO** as a topic.

### Teaching expansion

**Method Resolution Order** is the order Python follows when searching classes for an attribute or method, especially when multiple inheritance is used.

For a class with several inheritance paths, Python needs a predictable search order.

You can inspect it with:

```python
ClassName.mro()
```

````python
print(Child.mro())
````

## 7.4 Hierarchical Inheritance

**One parent → multiple children**

Two or more child classes inherit from the same parent class.

````python
class Animal:
    def eat(self):
        print("Animal can eat")


class Dog(Animal):
    def bark(self):
        print("Dog barks")


class Cat(Animal):
    def meow(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.eat()
dog.bark()

cat.eat()
cat.meow()
````

# 9. Polymorphism

**Polymorphism** means **"having more than one form."**

It allows the same method or operation to behave differently in different situations.

The PDF first gives the general OOP classification:

1. **Static polymorphism** — compile-time polymorphism
2. **Dynamic polymorphism** — run-time polymorphism

## 9.1 Static Polymorphism — general OOP theory

It associates compile-time polymorphism with:

- method overloading,
- operator overloading,
- selecting the matching operation at compile time.

The PDF states that Python does **not** support traditional compile-time method overloading in the same way because Python is dynamically typed and method resolution is not based on strict compile-time signatures.

A later method definition with the same name replaces an earlier one rather than creating a traditional overload set.

in java

````python
class Calculator {

    int add(int a, int b) {
        return a + b;
    }

    int add(int a, int b, int c) {
        return a + b + c;
    }
}
````

````python
Calculator c = new Calculator();

System.out.println(c.add(10, 20));
System.out.println(c.add(10, 20, 30));
````

Python does not support traditional method overloading based on method signatures like Java or C++.

````python
class Test:
    def add(self, a, b):
        return a + b

    # This definition replaces the previous method with the same name.
    def add(self, a, b, c):
        return a + b + c

t = Test()
print(t.add(1, 2))
````

````python
class Calculator:

    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c


c = Calculator()

print(c.add(10, 20))
````

Python does not support traditional compile-time method overloading directly. Similar behavior can be achieved using default arguments, *args, or other runtime techniques.
