## Introduction

Data does not always arrive in the exact type a program needs. Input from a user is always text, even if
it represents a number. **Type conversion** (also called **type casting**) means changing a value from
one type into another, equivalent type.


## Detailed Explanation

### 1. What type conversion is and why it is needed
Type conversion changes a value's type so it can be used correctly — for example, turning the text
`"25"` into the number `25`, so it can be used in math.


````python
age_text = "20"
age_number = int(age_text)
print(age_number, type(age_number))
````

**Observation:** The text `"20"` became the real number `20` after using `int()`.


### 2. Explicit and implicit conversion
**Explicit** conversion means the programmer does it on purpose, using `int()`, `float()`, `str()`, or
`bool()`. **Implicit** conversion happens automatically, without the programmer asking, mostly between
whole numbers and decimals during a calculation.


````python
# Explicit example
number_as_text = str(20)
print(type(number_as_text))

# Implicit example (Python converts automatically here)
result = 5 + 2.5
print(result, type(result))
````

**Observation:** In the first line, we deliberately called `str()`. In the second line, Python
silently turned the whole number into a decimal during the calculation, without us asking.


### 3. Converting values between numeric, text, and Boolean types


````python
print(int("100"))       # text to number
print(str(100))          # number to text
print(float(7))            # whole number to decimal
print(bool(1))              # number to True/False
````

**Observation:** Each conversion function changed the value into the exact type named by the
function.


### 4. When conversion succeeds and when it produces errors
Conversion succeeds when the value makes sense in the target type. It fails, raising a `ValueError`, when
it does not.


````python
print(int("42"))     # succeeds

try:
    print(int("hello"))     # fails
except ValueError as error:
    print("Error:", error)
````

**Observation:** `"42"` is a real number written as text, so conversion succeeds. `"hello"` is not a
number at all, so conversion fails with a `ValueError`.


### 5. How conversion affects calculations and comparisons


````python
a = "5"
b = "3"
print(a + b)     # joins text: "53"

a = int(a)
b = int(b)
print(a + b)      # adds numbers: 8
````

**Observation:** Same variables, same `+`, but converting to numbers first completely changed the
answer, from joined text to a real sum.


## Multiple Experiments

### Experiment 1: Converting a Decimal to a Whole Number

**What We Are Testing:** Whether `int()` rounds a decimal number or simply drops the decimal part.


````python
decimal = 9.8
as_int = int(decimal)
print(as_int)
````

**Observation:** `9.8` becomes `9`, NOT `10` — `int()` cuts off the decimal part instead of
rounding.


### Experiment 2: Trying int() on Decimal-Looking Text

**What We Are Testing:** Whether `int()` can directly convert text that contains a decimal point.


````python
try:
    value = int("7.5")
except ValueError as error:
    print("Error:", error)

# Correct way - convert to float first
value = float("7.5")
print(value)
````

**Observation:** `int()` fails on `"7.5"` because it contains a decimal point. `float()` handles it
correctly instead.


### Experiment 3: Converting Various Values to Boolean

**What We Are Testing:** Which values become `True` and which become `False` when converted with
`bool()`.


````python
print(bool(1))
print(bool(0))
print(bool(""))
print(bool("hello"))
print(bool([]))
````

**Observation:** `0`, an empty string, and an empty list all become `False`. Almost everything else
becomes `True`.


### Experiment 4: Comparing a Number With Its Text Version

**What We Are Testing:** Whether `5` and `"5"` are considered equal without conversion.


````python
print(5 == "5")
print(5 == int("5"))
````

**Observation:** Without conversion, `5 == "5"` is `False`, because they are different types. After
converting `"5"` with `int()`, the comparison correctly becomes `True`.


### Experiment 5: Adding Two Numbers Entered as Text

**What We Are Testing:** Whether converting two text values allows them to be added correctly.


````python
first_value = "12"
second_value = "8"

wrong_result = first_value + second_value
correct_result = int(first_value) + int(second_value)

print("Without conversion:", wrong_result)
print("With conversion:", correct_result)
````

**Observation:** Without conversion, `+` joined the text into `"128"`. After converting both to
numbers, `+` correctly added them to `20`.


### Experiment 6: Checking if bool("False") is Actually False

**What We Are Testing:** Whether the text `"False"` converts to the Boolean value `False`.


````python
print(bool("False"))
print(bool(""))
````

**Observation:** Surprisingly, `bool("False")` is `True`! Any non-empty string is considered `True`
in Python, no matter what words it contains. Only a truly EMPTY string (`""`) becomes `False`.


## Important Rules
- `int()`, `float()`, `str()`, `bool()` are the main explicit conversion tools.
- Converting a decimal to a whole number CUTS OFF the decimal part rather than rounding.
- `int()` cannot directly convert text containing a decimal point; use `float()` first if needed.
- Implicit conversion mostly happens automatically only between int and float during math.


## Common Mistakes

**Mistake 1: Assuming int() rounds a decimal**
```
int(9.8)   # gives 9, NOT 10
```
This is wrong if the beginner expects rounding — `int()` always truncates (cuts off) the decimal part.
**How to avoid it:** Use Python's `round()` function first if rounding is actually needed.

**Mistake 2: Using int() on decimal-looking text**
```
int("4.5")   # raises a ValueError
```
This is wrong because `int()` cannot handle a decimal point in the text.
**How to avoid it:** Use `float("4.5")` instead, then convert to `int()` afterward if truly needed.

**Mistake 3: Comparing a number to its text version**
```
5 == "5"   # this is False, not True
```
This is wrong if the beginner expects them to be treated as equal.
**How to avoid it:** Convert one side first, e.g. `5 == int("5")`.
