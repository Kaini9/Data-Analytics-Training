## Introduction

If a program never shows you anything, how would you know it worked? `print()` is Python's way of
showing text on your screen. It is one of the very first tools every beginner learns, and you will use it
constantly to display messages, results, and answers — from your very first program to advanced
projects later on.


## Detailed Explanation

### 1. The purpose of displaying output
Output lets a program communicate its results to a human. Without it, we could never see what a program
calculated or decided.


````python
print("This program is running.")
````

**Observation:** The message appeared on the screen because we used `print()` to display it.


### 2. Printing text, numbers, variables, and calculated results


````python
print("Hello, Ram!")           # text
print(42)                        # number
name = "Hari"
print(name)                       # variable
print(5 + 3)                       # calculated result
````

**Observation:** `print()` works the same no matter what kind of value it is given — text, a plain
number, a variable, or the answer to a calculation.


### 3. Displaying multiple values
You can pass several values to `print()`, separated by commas.


````python
name = "Sita"
marks = 90
print("Name:", name, "Marks:", marks)
````

**Observation:** All four items appeared on one line, automatically separated by spaces.


### 4. Output separators
By default, Python puts a single space between printed items. The `sep` parameter lets you change that.


````python
print("Ram", "Hari", "Shyam")               # default separator: a space
print("Ram", "Hari", "Shyam", sep=", ")       # custom separator
````

**Observation:** Changing `sep` only changed what appears BETWEEN the names — nothing changed before
the first name or after the last one.


### 5. Ending behavior
By default, `print()` ends with a new line. The `end` parameter lets you change what happens after
printing instead.


````python
print("Loading", end="...")
print("Done!")
````

**Observation:** Because `end="..."` replaced the usual new line, the second `print()` continued
right after the first one, on the same line.


### 6. How output formatting changes based on different settings
Combining `sep` and `end` lets you fully control how the output looks.


````python
print("x", "y", "z", sep=" -> ", end=" [done]\n")
print("next line starts here")
````

**Observation:** `sep` changed the spacing between `x`, `y`, and `z`, while `end` changed what
appeared right after `z`, before moving to the next line.


## Multiple Experiments

### Experiment 1: An Empty print()

**What We Are Testing:** What happens when `print()` is called with nothing inside it.


````python
print("Before")
print()
print("After")
````

**Observation:** Even with nothing inside it, `print()` still moved to a new line, creating a blank
line between "Before" and "After".


### Experiment 2: Changing sep to an Empty String

**What We Are Testing:** What happens when `sep` is set to nothing at all.


````python
print("A", "B", "C", sep="")
````

**Observation:** With `sep=""`, the letters are joined directly with no space between them at all.


### Experiment 3: Printing a Calculated Result Directly

**What We Are Testing:** Whether print() can show the result of a calculation without saving it in a
variable first.


````python
print(7 * 6)
````

**Observation:** Python calculated `7 * 6` first, and only the final answer (`42`) was shown — the
calculation itself was never visible.


### Experiment 4: Using sep to Print Each Item on Its Own Line

**What We Are Testing:** Whether `sep` can be used to create new lines between items in ONE print call.


````python
print("Apple", "Banana", "Cherry", sep="\n")
````

**Observation:** Setting `sep="\n"` placed each item on its own separate line, even though it was
all one single `print()` call.


### Experiment 5: Keeping Several print() Calls on the Same Line

**What We Are Testing:** Using `end` across multiple print() calls to build one combined line of output.


````python
print("5", end=" ")
print("4", end=" ")
print("3", end=" ")
print("2", end=" ")
print("1")
````

**Observation:** All five numbers appeared on ONE line because each `print()` (except the last) used
`end=" "` instead of the default new line.


### Experiment 6: Forgetting a Comma Between Items

**What We Are Testing:** What error Python gives when a comma is missing between two items in print().


````python
try:
    exec('print("Age:" 20)')
except SyntaxError as error:
    print("Error:", error)
````

**Observation:** Without a comma, Python cannot tell where one item ends and the next begins, so it
raises a `SyntaxError`.


## Important Rules
- The default separator between items is a single space.
- The default ending is a new line.
- Items in `print()` must be separated by commas.
- `sep` and `end` must be given with an `=` sign, and their values must be text.


## Common Mistakes

**Mistake 1: Forgetting the comma between items**
```
print("Age:" 20)
```
This is wrong because Python doesn't know how to separate the two items without a comma.
**How to avoid it:** Always place a comma between separate items: `print("Age:", 20)`.

**Mistake 2: Joining text and a number with + inside print**
```
print("Age: " + 20)
```
This is wrong because `+` cannot directly join text and a number — it causes a `TypeError`.
**How to avoid it:** Use a comma instead: `print("Age:", 20)`.

**Mistake 3: Forgetting sep and end need an = sign**
```
print("A", "B", sep ", ")
```
This is wrong because `sep` is a keyword argument and must be written as `sep=", "`.
**How to avoid it:** Always include the equals sign when using `sep` or `end`.
