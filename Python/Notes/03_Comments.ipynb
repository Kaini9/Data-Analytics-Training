## Introduction

A **comment** is a note written directly in your code that Python completely ignores when the program
runs — it exists purely for human readers. You write one by placing a `#` symbol before it.

Comments matter because code can become confusing later, even to the person who wrote it. A comment can
explain WHY something was done a certain way, which is often more useful than restating WHAT the code
already shows just by reading it. Beginners will use comments constantly to leave notes for themselves
and others.


## Detailed Explanation

### 1. What comments are and why they are useful
Comments explain the purpose or reasoning behind code — something the code itself often cannot show on
its own.


````python
# This comment explains the line below: it prints a welcome message
print("Welcome, Ram!")
````

**Observation:** The comment line did not affect the program at all — only the `print()` line
actually produced output.


### 2. Single-line explanations
Anything written after `#` on that line is ignored by Python, whether the comment is on its own line or
placed after real code.


````python
# This is a single-line comment on its own
print("Hari's score is 90.")   # this comment is at the end of a real line
````

**Observation:** Both comments — the one on its own line and the one after the `print()` — were
ignored, and only the message printed.


### 3. Multi-line explanations
Python has no special symbol just for multi-line comments. Programmers either repeat `#` on each line, or
use an unused triple-quoted string as a block-style note.


````python
# Step 1: set the price
# Step 2: set the quantity
# Step 3: calculate the total

price = 25
quantity = 4
total = price * quantity
print("Total:", total)
````

**Observation:** Three separate `#` lines together explained the whole block of code before it ran.


### 4. Using comments to document and explain programs
Good comments describe the PURPOSE of code, not just repeat what it obviously does.


````python
# This program calculates Sita's exam average
sita_marks = [88, 92, 79]
average = sum(sita_marks) / len(sita_marks)
print("Sita's average:", average)
````

**Observation:** The comment tells us WHY this code exists (to calculate an average), which is more
helpful than simply describing each line.


### 5. Using comments while testing or temporarily disabling statements
Placing `#` in front of REAL code stops that line from running — useful while testing a program.


````python
print("Step 1")
# print("Step 2 - turned off for now")
print("Step 3")
````

**Observation:** Because Step 2's line starts with `#`, it did NOT run — the output jumps straight
from Step 1 to Step 3. Removing the `#` later would bring it back exactly as before.


### 6. The difference between comments and normal text values
A `#` comment is thrown away by Python — it is never real data. A string (even an unused one) IS real
data, even if nothing ever uses it.


````python
# This is a comment - it shows NOTHING when run.

"This is just a plain string, sitting alone"

print("Only this line actually prints something on purpose.")
````

**Observation:** In a notebook, an unused string left alone on the LAST line of a cell is shown as
the cell's result (a notebook-specific behavior). A true `#` comment can NEVER show up like that — Python
throws it away completely, no matter where it is written.

**What We Are Testing:** What happens if you meant to write a comment but forgot the `#`.


````python
try:
    exec("This was supposed to be a comment")
except SyntaxError as error:
    print("Error:", error)
````

**Observation:** Without the `#`, Python tries to read the text as actual code and fails with a

**Observation:** The triple-quoted block acted like a note above the code, without affecting the

## Important Rules
- `#` comments only apply to the rest of THAT line, not the next line.
- A good comment explains WHY, not just WHAT (the code often already shows what).
- A comment is not a real value — it's thrown away by Python.
- Python has no dedicated multi-line comment symbol; repeated `#` lines are the common approach.


## Common Mistakes

**Mistake 1: Writing comments that just repeat the obvious**
```
x = 5  # set x to 5
```
This is wrong because the comment adds no new information beyond what the code already shows.
**How to avoid it:** Explain WHY, e.g. `x = 5  # starting score before the game begins`.

**Mistake 2: Forgetting a comment stops at the end of the line**
```
# This explains steps 1 and 2
step_one()
step_two()
```
This is wrong if the writer intended the comment to explain BOTH lines below it — Python only skips the
comment's own line, not the lines after it (those still run normally, which is fine, but it's a common
misunderstanding about how far a comment "reaches").
**How to avoid it:** Add a separate `#` comment above each line if each needs its own explanation.

**Mistake 3: Leaving old commented-out code lying around**
```
# print("old debug line")
# print("another old debug line")
print("actual working code")
```
This is wrong because it clutters the file and confuses future readers about what is still needed.
**How to avoid it:** Remove old commented-out code once it is no longer needed.
