## Introduction

A **variable** is like a labeled box where you store a value. You give the box a name, and later you can
look inside it (by using its name) to get the value back, or replace it with something new.

Variables matter because programs need to REMEMBER things — a person's name, a score, a total price.
Without variables, a program could not carry a value from one line of code to the next. You will use
variables in literally every program you write from now on.


## Detailed Explanation

### 1. What variables are and why they are used
A variable stores a value under a name so the program can reuse that value later, without retyping it.


````python
student_name = "Ram"
print(student_name)
````

**Observation:** We stored the text `"Ram"` inside the variable `student_name`, and printing the
variable shows us that value.


### 2. Variable creation and assignment
You create a variable by writing `name = value`. Python does not need you to state the type in advance —
it figures that out automatically.


````python
age = 20
print(age)
````

**Observation:** We didn't tell Python that `age` would hold a number — it worked that out just from
the value `20` we assigned.


### 3. Variable naming rules
- Can use letters, numbers, and underscores `_`.
- Cannot start with a number.
- Cannot contain spaces.
- Cannot be a Python reserved word (like `if`, `for`, `class`).


````python
student_age = 20
print(student_age)
````

**Observation:** `student_age` follows all the naming rules — letters and an underscore, no spaces, no
leading number.


### 4. Valid and invalid variable names


````python
# Valid names
hari_score = 90
_total = 100
score2 = 50
print(hari_score, _total, score2)

# Invalid names would cause errors if uncommented:
# 2total = 100      -> starts with a number
# my score = 90      -> contains a space
````

**Observation:** All three valid names ran without error. If we removed the `#` in front of the
invalid ones, Python would show a `SyntaxError`.


### 5. Case sensitivity
Python treats uppercase and lowercase letters as DIFFERENT. `age`, `Age`, and `AGE` are three separate
variables.


````python
age = 20
Age = 30
AGE = 40
print(age, Age, AGE)
````

**Observation:** Even though these three names look almost the same to us, Python stored three
completely separate values.


### 6. Storing different types of values
A variable can hold a whole number, a decimal, text, or a True/False value — Python figures out the type
automatically.


````python
name = "Sita"          # text
marks = 85               # whole number
percentage = 92.5         # decimal
passed = True              # True/False

print(name, marks, percentage, passed)
````

**Observation:** Four different kinds of values were stored, each in its own variable, with no extra
setup needed.


### 7. Multiple assignment and assigning one value to multiple variables
You can assign several variables in ONE line — either different values to each, or the SAME value to
all of them at once.


````python
# Different values in one line
x, y, z = 1, 2, 3
print(x, y, z)

# Same value assigned to multiple variables at once
a = b = c = 0
print(a, b, c)
````

**Observation:** In the first line, each variable got its own separate value. In the second line, all
three variables point to the exact same value, `0`.


### 8. Changing variable values
A variable is not permanent — reassigning it replaces the old value, and can even change its type.


````python
score = 0
print("Start:", score)

score = 10
print("Changed:", score)

score = "high score"   # even the TYPE can change
print("Changed again:", score)
````

**Observation:** The same variable name held a number, then later held text. Reassigning simply
replaces whatever was there before.


### 9. Checking the type of stored values
`type()` tells you what kind of value is currently stored inside a variable.


````python
marks = 85
print(type(marks))

name = "Hari"
print(type(name))
````

**Observation:** `type()` confirmed `marks` is a whole number (`int`) and `name` is text (`str`).


## Multiple Experiments

### Experiment 1: Trying an Invalid Variable Name

**What We Are Testing:** What error Python shows when a name starts with a digit.


````python
try:
    exec("1name = 'Ram'")
except SyntaxError as error:
    print("Error:", error)
````

**Observation:** Python refuses to accept a name that starts with a digit and shows a `SyntaxError`.


### Experiment 2: Case Sensitivity in Action

**What We Are Testing:** Whether `total` and `Total` are treated as the same variable.


````python
total = 100
Total = 200
print("total:", total)
print("Total:", Total)
````

**Observation:** They hold different values, proving Python sees them as two separate variables.


### Experiment 3: Reassigning a Variable to a Different Type

**What We Are Testing:** What happens to a variable's type when it is reassigned with a different kind of
value.


````python
value = 10
print(value, type(value))

value = "ten"
print(value, type(value))
````

**Observation:** The type changed from `int` to `str` simply because we assigned a different kind of
value. Python does not lock a variable to its first type.


### Experiment 4: Multiple Assignment with Different Values

**What We Are Testing:** Assigning three different exam marks to three variables in a single line.


````python
math_marks, science_marks, english_marks = 78, 85, 90
print(math_marks, science_marks, english_marks)
````

**Observation:** Each variable received its own value, matched by position, in one short line of code.


### Experiment 5: Assigning One Value to Multiple Variables

**What We Are Testing:** Giving several variables the SAME starting value at once.


````python
score1 = score2 = score3 = 0
print(score1, score2, score3)

score1 = 50
print(score1, score2, score3)
````

**Observation:** All three started at `0`. Changing only `score1` afterward did NOT change `score2` or
`score3` — each variable now holds its own independent value.


### Experiment 6: Checking Types of Several Values at Once

**What We Are Testing:** Using `type()` on a mix of different values to confirm Python identifies each
one correctly.


````python
values = [10, 3.5, "Shyam", True, None]
for v in values:
    print(v, "->", type(v))
````

**Observation:** Each value's type was identified correctly and automatically, just from how the value
was written.


## Important Rules
- A variable name cannot start with a number.
- A variable name cannot contain spaces.
- Uppercase and lowercase letters are treated as different names.
- A variable can be given a new value — and even a new TYPE of value — at any time.
- Python decides a variable's type automatically, based on its current value.


## Common Mistakes

**Mistake 1: Starting a name with a digit**
```
1st_place = "Ram"
```
This is wrong because Python requires names to start with a letter or underscore, not a digit.
**How to avoid it:** Rename it to something like `first_place`.

**Mistake 2: Assuming Total and total are the same variable**
```
Total = 100
print(total)   # this is a DIFFERENT variable, likely undefined
```
This is wrong because Python treats capitalization as meaningful — `Total` and `total` are unrelated.
**How to avoid it:** Pick one consistent capitalization style and stick to it throughout your program.

**Mistake 3: Using unclear variable names**
```
x1 = 90
d2 = "Ram"
```
This is wrong because these names don't describe what they hold, making the code hard to understand
later.
**How to avoid it:** Use descriptive names like `math_score` or `student_name`.
