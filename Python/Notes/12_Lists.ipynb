## Introduction

So far, one variable held ONE value. Real programs often need to work with GROUPS of related values — a
class list of student names, a set of exam scores, a shopping list. A **list** is Python's most commonly
used collection: an ordered, changeable sequence of values, all stored together in one variable.


## Detailed Explanation

### 1. What lists are and why collections are useful
A list groups multiple related values together under one name, so we can work with them as a single
unit instead of many separate variables.


````python
fruits = ["apple", "banana", "cherry"]
print(fruits)
````

**Observation:** Three separate fruit names are stored together in one variable, instead of needing
three separate variables.


### 2. Creating lists and storing multiple values
A list is created with square brackets `[]`, with values separated by commas.


````python
numbers = [10, 20, 30, 40]
print(numbers)
print(len(numbers))
````

**Observation:** `len()` correctly counted 4 items inside the list.


### 3. Storing different data types
A single list can freely mix different types of values together.


````python
mixed = ["Ram", 21, True, 5.5]
print(mixed)
````

**Observation:** Text, a whole number, a Boolean, and a decimal all live together happily in the
same list.


### 4. Accessing elements using positive and negative indexing
Just like strings, list elements are accessed starting at `0`. Negative indexing counts from the end,
with `-1` as the last element.


````python
fruits = ["apple", "banana", "cherry", "mango"]

print(fruits[0])     # first item
print(fruits[-1])     # last item
````

**Observation:** `fruits[0]` gave the first item, and `fruits[-1]` gave the last item, counted from
the opposite direction.


### 5. Slicing lists
A slice extracts a portion of a list, following the same rules as string slicing — the ending index is
excluded.


````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:4])
print(numbers[:3])
print(numbers[::-1])     # reversed
````

**Observation:** `numbers[1:4]` picked items from index 1 up to (but not including) index 4.


### 6. Changing existing elements
Unlike strings, lists ARE mutable — an existing element can be directly reassigned using its index.


````python
fruits = ["apple", "banana", "cherry"]
print("Before:", fruits)

fruits[1] = "blueberry"
print("After:", fruits)
````

**Observation:** The second item changed directly, in place, without needing to create a whole new
list.


### 7. Adding and removing elements
`.append(x)` adds `x` to the end. `.insert(pos, x)` adds `x` at a specific position. `.remove(x)` removes
the first item equal to `x`. `.pop()` removes and returns the last item (or a specific position).


````python
tasks = ["Wake up", "Study"]
print("Start:", tasks)

tasks.append("Exercise")
print("After append:", tasks)

tasks.insert(1, "Eat breakfast")
print("After insert:", tasks)

tasks.remove("Study")
print("After remove:", tasks)

removed_item = tasks.pop()
print("Popped item:", removed_item)
print("After pop:", tasks)
````

**Observation:** `.append()` always adds at the end, `.insert()` adds at a chosen position,
`.remove()` deletes by VALUE, and `.pop()` deletes by POSITION while also giving back the removed item.


### 8. Finding the length of a list
`len()` works on lists the same way it works on strings.


````python
students = ["Ram", "Hari", "Sita", "Shyam"]
print(len(students))
````

**Observation:** `len()` correctly counted 4 names in the list.


### 9. Sorting, reversing, searching, counting, and clearing list data


````python
scores = [88, 95, 72, 95, 60]

print("New sorted copy:", sorted(scores))
print("Original unchanged:", scores)

scores.sort()      # sorts the ORIGINAL list
print("After .sort():", scores)

scores.reverse()
print("After .reverse():", scores)

print("Index of 72:", scores.index(72))
print("Count of 95:", scores.count(95))

scores.clear()
print("After clear():", scores)
````

**Observation:** `sorted()` gave a new list and left the original alone, `.sort()` changed the
original list directly, and `.clear()` emptied it completely.


## Multiple Experiments

### Experiment 1: Real-World Collection - Student Marks

**What We Are Testing:** Using a list to store and analyze a group of related numbers.


````python
student_marks = [78, 85, 92, 66, 90]

print("Number of students:", len(student_marks))
print("Highest mark:", max(student_marks))
print("Lowest mark:", min(student_marks))
print("Average mark:", sum(student_marks) / len(student_marks))
````

**Observation:** Combining `len()`, `max()`, `min()`, and `sum()` with a list made it easy to
calculate simple statistics over a group of real data.


### Experiment 2: Changing Values in a Loop

**What We Are Testing:** Whether we can access and print each item in a list one at a time.


````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("I like", fruit)
````

**Observation:** The loop visited each item in the list in order, letting us use each one
individually.


### Experiment 3: Trying to Remove a Value That Doesn't Exist

**What We Are Testing:** What error occurs when `.remove()` is used with a value not in the list.


````python
fruits = ["apple", "banana"]
try:
    fruits.remove("mango")
except ValueError as error:
    print("Error:", error)
````

**Observation:** `.remove()` raises a `ValueError` when the value being removed does not exist in
the list.


### Experiment 4: Accessing an Index That Doesn't Exist

**What We Are Testing:** What error occurs when accessing a position beyond the list's length.


````python
fruits = ["apple", "banana"]
try:
    print(fruits[5])
except IndexError as error:
    print("Error:", error)
````

**Observation:** Just like with strings, asking for a position beyond the list's length raises an
`IndexError`.


### Experiment 5: Combining Two Lists

**What We Are Testing:** Whether `+` combines two lists into one, similar to how it joins two strings.


````python
list_a = [1, 2, 3]
list_b = [4, 5, 6]
combined = list_a + list_b
print(combined)
````

**Observation:** `+` combined both lists into one longer list, in the order they were written.


### Experiment 6: Names and Matching Scores Together

**What We Are Testing:** Using two related lists together, matched by their shared index positions.


````python
names = ["Ram", "Hari", "Sita"]
scores = [78, 85, 92]

for i in range(len(names)):
    print(names[i], "scored", scores[i])
````

**Observation:** Using the same index `i` for both lists let us match each name with its
corresponding score correctly.


## Important Rules
- Lists are ordered and mutable — elements can be changed, added, and removed after creation.
- List indexing and slicing follow the exact same rules as string indexing and slicing.
- `sorted()` returns a new list; `.sort()` modifies the original list in place and returns `None`.
- `.remove()` deletes by value; `.pop()` deletes by position (and returns the removed item).


## Common Mistakes

**Mistake 1: Expecting .sort() to return a new sorted list**
```
result = scores.sort()
print(result)   # this prints None, not the sorted list
```
This is wrong because `.sort()` changes the list in place and returns `None`.
**How to avoid it:** Just call `scores.sort()` on its own line, then print `scores` afterward.

**Mistake 2: Using .remove() with a value that doesn't exist**
```
fruits.remove("mango")   # error if "mango" is not in the list
```
This is wrong because `.remove()` needs to find a matching value first.
**How to avoid it:** Check with `if "mango" in fruits:` before removing it.

**Mistake 3: Accessing an index beyond the list's length**
```
fruits[10]   # error if the list has fewer than 11 items
```
This is wrong because that position simply does not exist.
**How to avoid it:** Check the list's length with `len()` before accessing a specific index.
