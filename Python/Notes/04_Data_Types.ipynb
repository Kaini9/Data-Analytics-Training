## Introduction

Every value in Python belongs to a **type** — a category that tells Python what KIND of thing it is
dealing with. A number can be added to another number, but adding a number directly to a word does not
make sense unless we first convert one of them.

Data types matter because what you can DO with a value depends entirely on its type. Beginners run into
type-related errors very often, so understanding this topic early will save a lot of confusion later.


## Detailed Explanation

### 1. What data types are and why they matter
A data type tells Python (and us) what kind of value we are working with, and which operations make
sense on it.


````python
age = 20
name = "Ram"
print(type(age), type(name))
````

**Observation:** `age` and `name` hold very different kinds of values, and `type()` confirms this —
one is a number, the other is text.


### 2. Numbers, decimal values, text, Boolean values, collections, mappings, and the absence of a
value
- **Numbers (int):** `5`, `-3`, `100`
- **Decimal values (float):** `3.14`, `-0.5`
- **Text (str):** `"hello"`, `"Ram"`
- **Boolean values:** `True` or `False`
- **Collections (list, set):** `[1, 2, 3]`, `{1, 2, 3}`
- **Mappings (dict):** `{"name": "Ram", "age": 20}`
- **Absence of a value (None):** `None`


````python
whole_number = 10
decimal_number = 3.14
text_value = "Hari"
true_or_false = True
a_list = [1, 2, 3]
a_set = {1, 2, 3}
a_dictionary = {"name": "Hari", "age": 22}
nothing = None

for value in [whole_number, decimal_number, text_value, true_or_false, a_list, a_set, a_dictionary, nothing]:
    print(value, "->", type(value))
````

**Observation:** Each of the seven kinds of values printed alongside its own type, showing how varied
Python's built-in types are.


### 3. How Python identifies the type of a value
Python looks at HOW a value was written: quotes mean text, `True`/`False` mean Boolean, square brackets
mean a list, and so on.


````python
print(type("20"))     # text, because of the quotes
print(type(20))         # a number, because there are no quotes
````

**Observation:** `"20"` and `20` look similar to a human, but the quotes make Python treat them as
completely different types.


### 4. Checking and comparing data types
Use `type(value)` to check a value's type, and `isinstance()` to check if a value belongs to a specific
type.


````python
marks = 85
print(type(marks) == int)
print(isinstance(marks, int))
print(isinstance(marks, str))
````

**Observation:** `marks` is confirmed to be an `int`, and both `type(marks) == int` and
`isinstance(marks, int)` return `True`, while `isinstance(marks, str)` correctly returns `False`.


### 5. How the same operation can behave differently depending on the data type
The `+` symbol changes meaning depending on the type of values used with it.


````python
print(5 + 3)               # adds numbers: 8
print("5" + "3")             # joins text: "53"
print([1, 2] + [3, 4])       # combines lists: [1, 2, 3, 4]
````

**Observation:** `5 + 3` gives `8` (a real math answer), but `"5" + "3"` gives `"53"` (text stuck
together). The exact same `+` symbol behaves completely differently depending on the type involved.


## Multiple Experiments

### Experiment 1: Mixing Text and a Number With +

**What We Are Testing:** What happens when we try to join text and a number directly with `+`.


````python
try:
    result = "Age: " + 25
except TypeError as error:
    print("Error:", error)
````

**Observation:** Python refuses to guess whether we meant math or joining text, and raises a
`TypeError` instead.


### Experiment 2: True and False Behaving Like 1 and 0

**What We Are Testing:** Whether Boolean values act like numbers in a calculation.


````python
print(True + True)
print(True == 1)
print(False == 0)
````

**Observation:** `True` behaves like `1` and `False` behaves like `0` in Python, which can be
surprising the first time you see it.


### Experiment 3: Comparing None to Other "Empty" Values

**What We Are Testing:** Whether `None`, `0`, and an empty string are treated as equal.


````python
nothing = None
print(nothing == 0)
print(nothing == "")
print(nothing == False)
````

**Observation:** `None` is not equal to `0`, `""`, or `False` — it is its own distinct value meaning
"nothing has been given here."


### Experiment 4: Changing the Values in a List and Checking Types Again

**What We Are Testing:** Whether changing what's inside a collection changes how Python reports its type.


````python
mixed_list = [1, "two", 3.0, True]
print(type(mixed_list))
for item in mixed_list:
    print(item, "->", type(item))
````

**Observation:** The list itself is always type `list`, no matter what mixture of types it contains
inside.


### Experiment 5: Comparing a Number and Its Text Version

**What We Are Testing:** Whether `5` and `"5"` are treated as equal by Python.


````python
print(5 == "5")
print(5 == 5.0)
````

**Observation:** A number and its text version are NEVER considered equal, but a whole number and an
equivalent decimal (`5` and `5.0`) ARE considered equal.


### Experiment 6: Using + on Sets Instead of Lists

**What We Are Testing:** Whether `+` works the same way on sets as it does on lists.


````python
try:
    result = {1, 2} + {3, 4}
except TypeError as error:
    print("Error:", error)
````

**Observation:** Unlike lists, sets do not support `+` at all — this shows that even collection types
can behave differently from each other.


## Important Rules
- Every value has exactly one type — check it with `type()`.
- `+` means "add" for numbers, but "join" for text, and "combine" for lists.
- `None` is different from `0`, `False`, and empty text/lists.
- `True` and `False` behave like `1` and `0` in numeric contexts.


## Common Mistakes

**Mistake 1: Adding text and a number directly**
```
"Age: " + 25
```
This is wrong because `+` cannot combine text and a number without conversion.
**How to avoid it:** Convert the number to text first (covered in the Type Casting notebook), or use a
comma inside `print()`.

**Mistake 2: Treating None, 0, False, and "" as interchangeable**
```
if value == None or value == 0 or value == False:
```
This is wrong because these are four DIFFERENT values, even though they may all seem "empty" in a loose
sense.
**How to avoid it:** Be specific about which exact value you are checking for.

**Mistake 3: Assuming user input is already a number**
```
age = input("Enter your age: ")
print(age + 1)   # this will error, because age is text
```
This is wrong because `input()` always returns text, even if it looks like a number.
**How to avoid it:** Convert with `int()` or `float()` before doing math (covered in later notebooks).
