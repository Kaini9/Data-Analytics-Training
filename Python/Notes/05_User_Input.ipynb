## Introduction

So far, our programs have used fixed values written directly into the code. Real programs usually need to
react to what a REAL PERSON tells them — like a quiz asking for an answer, or a calculator asking for two
numbers.

`input()` is Python's way of pausing the program, asking a question, and waiting for the person to type
something and press Enter. This is the first step toward building truly interactive programs.

**Note:** Running an `input()` cell in this notebook will show a small text box. Type an answer and press
Enter to continue.


## Detailed Explanation

### 1. What user input is and why interactive programs need it
User input lets a program get real information from the person using it, instead of always working with
fixed values written into the code.


````python
favorite_color = input("What is your favorite color? ")
print("Nice choice:", favorite_color)
````

**Observation:** Whatever we typed into the box was captured and used inside the program.


### 2. Accepting information from the user
`input()` shows a prompt message and waits for the person to type and press Enter.


````python
age = input("How old are you? ")
print("You said you are", age)
````

**Observation:** The text inside `input("...")` is the prompt message shown to the user before they
type anything.


### 3. Storing user input in a variable
Whatever is typed can be saved into a variable, just like any other value, and reused later in the
program.


````python
city = input("Which city do you live in? ")
print("City stored:", city)
````

**Observation:** The variable `city` now holds whatever the user typed, and we can use it again
anywhere later in the program.


### 4. The default type returned by user input
No matter what is typed, `input()` ALWAYS returns TEXT (a string) — even if it looks like a number.


````python
typed_value = input("Type a number: ")
print("You typed:", typed_value)
print("Its type is:", type(typed_value))
````

**Observation:** Even though you typed digits, Python stored it as `str` (text), NOT as a number.


### 5. Converting user input into appropriate numeric types
To do math, convert the text using `int()` (whole numbers) or `float()` (decimals).


````python
typed_value = input("Type a whole number: ")
number_value = int(typed_value)
print("Doubled:", number_value * 2)
````

````python
typed_value = input("Type a decimal number (example: 3.5): ")
decimal_value = float(typed_value)
print("Stored as decimal:", decimal_value, type(decimal_value))
````

**Observation:** After converting with `int()` or `float()`, the value became a real number, and math
worked correctly on it.


### 6. Handling common input-related mistakes
If input does not convert cleanly to a number, Python raises a `ValueError`.


````python
try:
    typed_value = input("Type a number: ")
    number_value = int(typed_value)
    print("Success:", number_value)
except ValueError as error:
    print("This wasn't a valid number:", error)
````

**Observation:** If the input can't be converted, the program shows a clear error message instead of
crashing unexpectedly.


## Multiple Experiments

### Experiment 1: Doing Math Without Converting First

**What We Are Testing:** What happens if we try to multiply raw (unconverted) input.


````python
typed_value = input("Type a number to double: ")
doubled = typed_value * 2
print("Result:", doubled)
````

**Observation:** If you typed `5`, you probably expected `10` — but you got `"55"` instead. That's
because `*` on TEXT means "repeat this text", not "multiply the number".


### Experiment 2: Fixing Experiment 1 by Converting First

**What We Are Testing:** Whether converting input to a number fixes the earlier problem.


````python
typed_value = input("Type a number to double: ")
number_value = int(typed_value)
doubled = number_value * 2
print("Correct result:", doubled)
````

**Observation:** Now the result is a real doubled number, because we converted the text to an
integer before multiplying.


### Experiment 3: Typing Letters Instead of a Number

**What We Are Testing:** What error occurs when the user types something that is not a valid number.


````python
try:
    typed_value = input("Type a whole number: ")
    number_value = int(typed_value)
    print("Converted:", number_value)
except ValueError as error:
    print("Error:", error)
````

**Observation:** Typing letters instead of digits causes a `ValueError`, since `int()` cannot make
sense of non-numeric text.


### Experiment 4: Comparing Two Numbers Entered by the User

**What We Are Testing:** Whether two separately entered numbers can be compared correctly after
conversion.


````python
first_number = int(input("Enter the first number: "))
second_number = int(input("Enter the second number: "))

if first_number > second_number:
    print("The first number is bigger.")
else:
    print("The second number is bigger or they are equal.")
````

**Observation:** Because both values were converted to real numbers, the comparison worked correctly
and gave a sensible answer.


### Experiment 5: Using Input to Build a Sentence

**What We Are Testing:** Combining several pieces of user input into one final message.


````python
name = input("What is your name? ")
city = input("Which city are you from? ")
print(name, "is from", city + ".")
````

**Observation:** Two separate pieces of user input were combined smoothly into one readable sentence.


### Experiment 6: Converting a Decimal Input for a Calculation

**What We Are Testing:** Whether a decimal input can be used correctly in a real calculation, like
calculating a total price.


````python
price = float(input("Enter the price of one item: "))
quantity = int(input("Enter the quantity: "))
total = price * quantity
print("Total cost:", total)
````

**Observation:** Converting `price` to `float` and `quantity` to `int` let us multiply them together
correctly to get a real total.


## Important Rules
- `input()` ALWAYS returns text — even if the person typed digits.
- To do math, convert with `int()` or `float()` first.
- `int()` cannot convert text with a decimal point (like `"3.5"`) — use `float()` for that.
- A prompt message inside `input("...")` is optional but strongly recommended.


## Common Mistakes

**Mistake 1: Doing math directly on raw input**
```
value = input("Enter a number: ")
print(value * 2)
```
This is wrong because `value` is text, so `* 2` repeats the text instead of multiplying it.
**How to avoid it:** Convert first: `int(value) * 2`.

**Mistake 2: Assuming digits typed by the user are automatically a number**
```
age = input("Enter your age: ")
print(type(age))   # this is str, not int
```
This is wrong because `input()` always gives text, regardless of what was typed.
**How to avoid it:** Always convert with `int()` or `float()` before treating input as a number.

**Mistake 3: Using int() on decimal-looking input**
```
value = int(input("Enter a decimal: "))   # fails if user types 3.5
```
This is wrong because `int()` cannot handle text with a decimal point.
**How to avoid it:** Use `float()` instead when decimal values are expected.
