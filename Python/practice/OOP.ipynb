```python
class Car:

    def __init__(self, name, color):
        self.name = name
        self.color = color

    def display(self):
        print("car name:", self.name)
        print("car color:", self.color)

car1 = Car("BMW", "Black")
car2 = Car("toyota", "White")
car1.display()
car2.display()
```
