```python
random.seed(42)
all_names = list({f"{first} {last}" for first in first_names for last in last_names})
random.shuffle(all_names)
names = all_names[:1050]

df = pd.DataFrame({
    "id": range(1, 1051),
    "name": names,
    "salary": [random.randint(30000, 150000) for _ in range(1050)],
    "job_title": [random.choice(job_titles) for _ in range(1050)]
})

df.to_csv("company_data_1050.csv", index=False)
```
