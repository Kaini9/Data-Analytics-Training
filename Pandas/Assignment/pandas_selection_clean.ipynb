# Pandas Data Selection — `[ ]`, `.loc`, `.iloc`
**Prepared by Ananda Rimal** · Dataset: Titanic

In this notebook we learn the 3 ways to select data from a DataFrame:

| Method | Selects by | Example |
|--------|-----------|---------|
| `df["col"]` | column name | `df["Age"]` |
| `df.loc[...]` | **labels** (names) | `df.loc[0:4, ["Name","Age"]]` |
| `df.iloc[...]` | **positions** (numbers) | `df.iloc[0:5, 0:3]` |

```python
import pandas as pd

df = pd.read_csv("/content/titanic_dataset.csv")   # change the path if needed
df.head()
```

## 1. Selecting Columns

```python
# Single column -> returns a SERIES (1 column)
df["Age"]
```

```python
# List of columns -> returns a DATAFRAME (note the double brackets!)
df[["Name", "Sex", "Age"]]
```

## 2. `.loc` — Label-based Selection (VERY IMPORTANT)

Syntax: `df.loc[row_label, column_label]`

`.loc` uses the **index labels** and column **names**.

```python
# Rows with labels 0 to 4  (loc INCLUDES the end -> 5 rows!)
df.loc[0:4]
```

```python
# Rows + specific columns together
df.loc[0:9, ["Name", "Age", "Sex"]]
```

```python
# One row, selected columns
df.loc[0, ["Name", "Age"]]
```

```python
# ALL rows, specific columns  (the ':' means 'everything')
df.loc[:, ["Survived", "Pclass", "Fare"]]
```

## 3. `.iloc` — Position-based Selection

Syntax: `df.iloc[row_position, column_position]`

`.iloc` uses **integer positions** (0, 1, 2, ...) — like a list.

```python
# First 5 rows  (iloc EXCLUDES the end -> positions 0,1,2,3,4)
df.iloc[0:5]
```

```python
# Rows AND columns by position: first 5 rows, first 3 columns
df.iloc[0:5, 0:3]
```

## 4. The Key Difference: `loc` vs `iloc`

| | `.loc[0:4]` | `.iloc[0:4]` |
|---|---|---|
| Selects by | label | position |
| End included? | **YES** (5 rows) | **NO** (4 rows) |

**Tip: GENERALLY USE `.loc` INSTEAD OF `.iloc`** — labels do not change when the data is
sorted or filtered, positions do.

## 5. `set_index` — Now Labels and Positions Are DIFFERENT

Until now the index labels (0,1,2,...) happened to equal the positions, so `loc` and `iloc`
looked the same. Watch what happens when we make **PassengerId** the index.

```python
df.set_index("PassengerId", inplace=True)
df.head()
```

```python
# Now the index labels are 1...891 (PassengerId), but positions are still 0...890

df.loc[1]       # LABEL 1  -> passenger with PassengerId = 1  (the FIRST row)
```

```python
df.iloc[887]    # POSITION 887 -> the 888th row (its PassengerId label is 888)
```

```python
# Same passenger, two ways:
print("By label   ->  df.loc[888]")
print("By position ->  df.iloc[887]")
df.loc[888].equals(df.iloc[887])
```

## Summary

- `df["col"]` -> one column (Series); `df[["c1","c2"]]` -> many columns (DataFrame)
- `.loc[rows, cols]` -> by **labels**, end **included**
- `.iloc[rows, cols]` -> by **positions**, end **excluded**
- After `set_index`, labels and positions become different — `.loc` follows the new index
- Prefer `.loc` in daily work
