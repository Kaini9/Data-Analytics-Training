LETS IMPORT THE PANDAS AND THE DATASET

```python
import pandas as pd
```

```python
df=pd.read_csv("/content/titanic_dataset.csv")
df.sample(2)
```

```python
df.info()
```

Q1. Count the total number of passengers in the Titanic dataset.


```python
df["PassengerId"].count()
```

### Q2. Count how many male and female passengers are present.


```python
df.groupby("Sex").size()
```

### Q3. Count how many passengers belong to each `Pclass`

```python
df.groupby("Pclass").size()
```

## Q4. Count how many passengers survived and how many died.

```python
df.groupby("Survived").size()
```

### Q5. Count how many passengers have missing `Age` values.



```python
df["Age"].isnull().sum()
```

### Q6. Count how many passengers have missing `Cabin` values.



```python
df["Cabin"].isnull().sum()
```


### Q7. Count how many passengers are male and survived.



shape[0], it counts row of selected column and conditions

```python
df.loc[(df["Sex"]=="male") & (df["Survived"]==1)].shape[0]
```

### Q8. Count the total number of passengers for each `Embarked` value.




```python
df.groupby("Embarked").size()
```


### Q9. Count passengers for every `Pclass` and `Sex` combination using `groupby().size()`.




```python
df.groupby(["Pclass","Sex"]).size()
```

### Q10. Create a cross-tabulation showing `Sex` against `Survived`.


```python
pd.crosstab(df["Sex"], df["Survived"])
```
