```python
import pandas as pd
```

lets create a dummy dataset to practice adding and removing

```python
#sample dataset creation
data = {
    "Name": ["Saval", "Nischal", "Manil", "Anush"],
    "Age": [22, 19, 25, 20],
    "Gender": ["male", "Male", "male", "male"],
    "Survived": [1, 1, 0, 1],
    "Fare": [50, 100, 30, 60]
}

df = pd.DataFrame(data)
df
```

SO now lets try to add a new column, lets add a column named ADDRESS

```python
df.loc[:,"Address"] = ["Nepal","Australia","UAE","USA"]
```

```python
df
```

lets manupulate the age 5 years back

```python
df.loc[:,"Age_five_yrs_before"]=df["Age"]-5
```

```python
df
```

Lets add adress 5 yr ago where we all were

```python
df.loc[:,"Address_fiveyr_ago"]=["Nepal","Nepal","Nepal","Nepal"]
```

```python
df
```

lets add status of all at once as all of us are single



```python
df.loc[:, "Status"]= "Single"
```

```python
df
```

By mistakenly i added in rows inplace of making new column, So lets remove it by its index using iloc

```python
df = df.iloc[:-1]
df
```

lets understand about .loc and inplace=true

.loc = WHERE / WHICH data

inplace=True = WHETHER to change the original DataFrame

.loc → selects a specific row/column and can change its value.

inplace=True → makes a pandas operation change the original DataFrame.

Sorting → use inplace=True if you want the sorted order saved directly in df.
