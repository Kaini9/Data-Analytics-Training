```python
import pandas as pd
```

```python
df= pd.read_csv('/content/big4_financial_risk_compliance.csv')
df.head(10)
```

```python
df.info()
```

```python
df.describe()
```

kun company ma employee workload dherai xata ra kun year ma

```python
df.loc[df["Employee_Workload"].max()== df["Employee_Workload"], ["Firm_Name","Employee_Workload","Year"]]
```

kun company ko 2024 ma workload 60 above ra highrisk cases dherai thyo

```python
df.loc[(df["Year"]== 2024) & (df["Employee_Workload"]>60 ), ["Firm_Name","Employee_Workload","Year","High_Risk_Cases"]]
```

100 vanda badi total revenue impact vako company ko name only in 2024 ma

```python
df.loc[(df["Year"]== 2024) & (df["Total_Revenue_Impact"]> 100), ["Firm_Name","Total_Revenue_Impact","Year"]]
```

Deloitte ko matra data herum

```python
df.loc[df["Firm_Name"]=="Deloitte", :]
```

deloitee ko data tara only 2020 ko matra herum hai ta

```python
df.loc[(df["Firm_Name"]=="Deloitte") & (df["Year"] == 2020),:]
```

healthcare affect vako company 2022 ma

```python
df.loc[(df["Industry_Affected"]=="Healthcare") & (df["Year"] == 2020),"Firm_Name"]
```

15 vanda badi fraud cases vako company in 2020

```python
df.loc[(df["Year"]== 2020) & (df["Fraud_Cases_Detected"]> 15), ["Firm_Name","Fraud_Cases_Detected","Year"]]
```

2020 ma kun ma thorai risk thyo






```python
df.loc[df["Year"]== 2020, ["Firm_Name","High_Risk_Cases","Year"]].min()
```

maximun chai kaslai thyo ta

```python
df.loc[df["Year"]== 2020, ["Firm_Name","High_Risk_Cases","Year"]].max()
```

kun company lai kun barsa dherai risk

```python
df.loc[df.groupby("Year")["High_Risk_Cases"].idxmax(),["Year", "Firm_Name", "High_Risk_Cases"]].sort_values("Firm_Name")
```

kun barsa ma kun firm lai risk dherai?

```python
df.loc[df.groupby("Year")["High_Risk_Cases"].idxmax(),["Year", "Firm_Name", "High_Risk_Cases"]].sort_values("Year")
```

2020 ma client satisfaction kasko badi

```python
df.loc[df["Year"] ==2020 ,["Firm_Name","Client_Satisfaction_Score","Year"]].max()
```

satisfaction kun ko max hola ta sabai barsa ma best

```python
df.loc[df["Client_Satisfaction_Score"].idxmax(),["Year", "Firm_Name", "Client_Satisfaction_Score"]]
```
